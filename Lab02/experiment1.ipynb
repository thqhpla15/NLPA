# Experiment 1: Corpus Statistics

This experiment analyzes the corpus and builds unigram, bigram,
and trigram language models.

The main objectives are:
- Compute corpus statistics.
- Build unigram, bigram, and trigram models.
- Analyze n-gram frequencies.
- Identify rare and unseen n-grams.

```python
# PHASE 1: SETUP

import json
import gzip
from collections import Counter

from ngram_lm import NGramLanguageModel
```

```python
# PHASE 2: LOAD CORPUS

CORPUS_PATH = "C:/Users/ASUS/Downloads/c4-train.00000-of-01024-30K.json.gz"

documents = []

with gzip.open(CORPUS_PATH, "rt", encoding="utf-8") as file:
    for line in file:
        documents.append(json.loads(line))

print("Number of documents:", len(documents))
```

```python
# PHASE 3: CORPUS STATISTICS

def tokenize(text):
    text = text.lower()

    tokens = []
    current = []

    for ch in text:
        if ch.isalpha():
            current.append(ch)
        else:
            if current:
                tokens.append("".join(current))
                current = []

    if current:
        tokens.append("".join(current))

    return tokens
```

```python
# Tokenize all documents

tokenized_documents = []

for document in documents:
    tokens = tokenize(document["text"])
    tokenized_documents.append(tokens)

total_documents = len(tokenized_documents)
total_tokens = sum(len(tokens) for tokens in tokenized_documents)

vocabulary = set()

for tokens in tokenized_documents:
    vocabulary.update(tokens)

vocabulary_size = len(vocabulary)

print("Number of documents:", total_documents)
print("Number of tokens:", total_tokens)
print("Vocabulary size:", vocabulary_size)
```

```python
# PHASE 4: BUILD N-GRAM MODELS

from ngram_lm import NGramLanguageModel

unigram_model = NGramLanguageModel(1)
bigram_model = NGramLanguageModel(2)
trigram_model = NGramLanguageModel(3)

unigram_model.fit(tokenized_documents)
bigram_model.fit(tokenized_documents)
trigram_model.fit(tokenized_documents)

print("Unigram model trained.")
print("Bigram model trained.")
print("Trigram model trained.")
```

```python
# Check basic model statistics

print("Vocabulary size:", len(unigram_model.vocabulary))
print("Total tokens:", unigram_model.total_tokens)

print("Unique unigrams:", len(unigram_model.unigram_counts))
print("Unique bigrams:", len(bigram_model.bigram_counts))
print("Unique trigrams:", len(trigram_model.trigram_counts))
```

```python
# PHASE 5: N-GRAM STATISTICS

TOP_K = 20

top_unigrams = unigram_model.unigram_counts.most_common(TOP_K)

print("Top", TOP_K, "Unigrams:")
for ngram, count in top_unigrams:
    print(f"{ngram}: {count}")
```

```python
top_bigrams = bigram_model.bigram_counts.most_common(TOP_K)

print("Top", TOP_K, "Bigrams:")
for ngram, count in top_bigrams:
    print(f"{ngram}: {count}")
```

```python
top_trigrams = trigram_model.trigram_counts.most_common(TOP_K)

print("Top", TOP_K, "Trigrams:")
for ngram, count in top_trigrams:
    print(f"{ngram}: {count}")
```

```python
# Count n-grams that occur exactly once

singleton_unigrams = sum(
    1 for count in unigram_model.unigram_counts.values()
    if count == 1
)

singleton_bigrams = sum(
    1 for count in bigram_model.bigram_counts.values()
    if count == 1
)

singleton_trigrams = sum(
    1 for count in trigram_model.trigram_counts.values()
    if count == 1
)

print("Unigrams occurring exactly once:", singleton_unigrams)
print("Bigrams occurring exactly once:", singleton_bigrams)
print("Trigrams occurring exactly once:", singleton_trigrams)
```

```python
# PHASE 6: RESULTS

results = {
    "Number of documents": total_documents,
    "Number of tokens": total_tokens,
    "Vocabulary size": vocabulary_size,
    "Unique unigrams": len(unigram_model.unigram_counts),
    "Unique bigrams": len(bigram_model.bigram_counts),
    "Unique trigrams": len(trigram_model.trigram_counts),
    "Singleton unigrams": singleton_unigrams,
    "Singleton bigrams": singleton_bigrams,
    "Singleton trigrams": singleton_trigrams
}

for metric, value in results.items():
    print(f"{metric}: {value}")
```

```python
# Display results as a simple table

print(f"{'Metric':<30} {'Value':>15}")
for metric, value in results.items():
    print(f"{metric:<30} {value:>15,}")
```

```python
import csv

with open("experiment1_results.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Metric", "Value"])

    for metric, value in results.items():
        writer.writerow([metric, value])
```
