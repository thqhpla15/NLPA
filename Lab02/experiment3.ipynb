# Experiment 3: Next-Word Prediction

This experiment evaluates next-word prediction using n-gram
language models.

The experiment investigates:
- Candidate next words for a given context.
- Next-word probabilities.
- Top-k predictions.
- The difference between bigram and trigram prediction.

```python
# PHASE 1: SETUP

import gzip
import json

from ngram_lm import NGramLanguageModel
```

```python
# PHASE 2: LOAD CORPUS

CORPUS_PATH = "C:/Users/ASUS/Downloads/c4-train.00000-of-01024-30K.json.gz"

documents = []

with gzip.open(CORPUS_PATH, "rt", encoding="utf-8") as file:
    for line in file:
        documents.append(json.loads(line))

print("Number of documents:", len(documents))
```

```python
# Tokenization

def tokenize(text):
    text = text.lower()

    tokens = []
    current = []

    for ch in text:
        if ch.isalpha():
            current.append(ch)
        else:
            if current:
                tokens.append("".join(current))
                current = []

    if current:
        tokens.append("".join(current))

    return tokens


tokenized_documents = []

for document in documents:
    tokens = tokenize(document["text"])
    tokenized_documents.append(tokens)

print("Number of documents:", len(tokenized_documents))
print("Number of tokens:", sum(len(tokens) for tokens in tokenized_documents))
```

```python
# PHASE 3: TRAIN N-GRAM MODELS

bigram_model = NGramLanguageModel(2)
trigram_model = NGramLanguageModel(3)

bigram_model.fit(tokenized_documents)
trigram_model.fit(tokenized_documents)

print("Bigram model trained.")
print("Trigram model trained.")
```

```python
# PHASE 4: BIGRAM NEXT-WORD PREDICTION

context = ["the"]

predictions = bigram_model.next_word(context)

print("Context:", context)
print("Number of predictions:", len(predictions))

for word, probability in predictions[:10]:
    print(f"{word}: {probability}")
```

```python
# PHASE 5: TOP-K PREDICTIONS

TOP_K = 10

top_predictions = predictions[:TOP_K]

print("Context:", context)
print()
print(f"Top {TOP_K} next-word predictions:")

for rank, (word, probability) in enumerate(
        top_predictions,
        start=1
):
    print(
        f"{rank}. {word:<20} "
        f"P = {probability:.8f}"
    )
```

```python
# PHASE 6: TRIGRAM NEXT-WORD PREDICTION

trigram_context = ["in", "the"]

trigram_predictions = trigram_model.next_word(
    trigram_context
)

print("Context:", trigram_context)
print("Number of predictions:", len(trigram_predictions))

print()
print("Top 10 predictions:")

for rank, (word, probability) in enumerate(
        trigram_predictions[:10],
        start=1
):
    print(
        f"{rank}. {word:<20} "
        f"P = {probability:.8f}"
    )
```

```python
# PHASE 7: BIGRAM VS TRIGRAM COMPARISON
bigram_context = ["the"]
trigram_context = ["in", "the"]

bigram_predictions = bigram_model.next_word(
    bigram_context
)

trigram_predictions = trigram_model.next_word(
    trigram_context
)

print("=" * 60)
print("BIGRAM")
print("=" * 60)

print("Context:", bigram_context)

for rank, (word, probability) in enumerate(
        bigram_predictions[:10],
        start=1
):
    print(
        f"{rank}. {word:<20} "
        f"P = {probability:.8f}"
    )

print()
print("=" * 60)
print("TRIGRAM")
print("=" * 60)

print("Context:", trigram_context)

for rank, (word, probability) in enumerate(
        trigram_predictions[:10],
        start=1
):
    print(
        f"{rank}. {word:<20} "
        f"P = {probability:.8f}"
    )
```

```python
# PHASE 8: RESULTS

print("=" * 60)
print("EXPERIMENT 3 RESULTS")
print("=" * 60)

print()
print("Bigram context:", bigram_context)
print("Top 10 predictions:")

for rank, (word, probability) in enumerate(
        bigram_predictions[:10],
        start=1
):
    print(
        f"{rank}. {word:<20} "
        f"P = {probability:.8f}"
    )

print()
print("Trigram context:", trigram_context)
print("Top 10 predictions:")

for rank, (word, probability) in enumerate(
        trigram_predictions[:10],
        start=1
):
    print(
        f"{rank}. {word:<20} "
        f"P = {probability:.8f}"
    )

print()
print("=" * 60)
print("Experiment 3 completed.")
print("=" * 60)
```
