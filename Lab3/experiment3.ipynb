```python
# Experiment 3: Effect of Context Window

# Create a corpus containing the word pairs required for the experiment
corpus_exp3 = [
    "the doctor works at the hospital",
    "the physician works at the hospital",
    "the doctor treats the patient",
    "the physician treats the patient",
    "the patient visits the hospital",
    "the doctor studies disease",
    "the physician studies disease",
    "the cat eats fish",
    "the dog eats fish",
    "the cat likes milk",
    "the dog likes meat"
]

# Convert each sentence into a list of words
sentences_exp3 = [sentence.split() for sentence in corpus_exp3]
```

```python
from gensim.models import Word2Vec

# Test different context window sizes
windows = [2, 5, 10]

# Store one model for each window size
models_exp3 = {}

for window in windows:
    model_exp3 = Word2Vec(
        sentences=sentences_exp3,
        vector_size=100,
        window=window,
        min_count=1,
        epochs=10
    )

    models_exp3[window] = model_exp3
```

```python
# Define the word pairs required by Experiment 3
word_pairs = [
    ("doctor", "physician"),
    ("doctor", "hospital"),
    ("cat", "dog")
]

# Calculate cosine similarity for each word pair and window size
for window in windows:
    model = models_exp3[window]

    print(f"\nWindow = {window}")

    for word1, word2 in word_pairs:
        similarity = model.wv.similarity(word1, word2)

        print(f"{word1} - {word2}: {similarity:.4f}")
```

**Is a larger window always better? (việc tăng kích thước cửa sổ ngữ cảnh lúc nào cũng tốt hơn không?)**

Độ tương đồng giữa doctor và physician vẫn ở khoảng 0,17, trong khi doctor–hospital vẫn gần -0,045 và cat–dog vẫn gần 0. Điều này cho thấy cửa sổ ngữ cảnh lớn hơn không phải lúc nào cũng tốt hơn.
