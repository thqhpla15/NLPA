## 1. Comparison of Word Representations

**Question:** How do TF-IDF, co-occurrence matrices, Word2Vec, and contextual embeddings differ in terms of representation, strengths, and limitations?

(TF-IDF, ma trận đồng xuất hiện, Word2Vec và contextual embeddings khác nhau như thế nào về cách biểu diễn, ưu điểm và hạn chế?)


| Representation | Type | Strengths | Limitations |
|---|---|---|---|
| **TF-IDF** | Sparse | Tốt cho **keyword matching** và **document retrieval** | Khó thể hiện quan hệ ngữ nghĩa giữa các từ |
| **Co-occurrence Matrix** | Sparse | Biểu diễn trực tiếp thông tin về **context** | Vector lớn, sparse và phụ thuộc vào **corpus** và **window size** |
| **Word2Vec** | Dense | Có khả năng thể hiện **semantic similarity** và quan hệ giữa các từ | Mỗi từ chỉ có một **static vector** |
| **Contextual Embedding** | Dense | Vector phụ thuộc vào **context**, xử lý tốt hơn các từ có nhiều nghĩa | Phức tạp và tốn **computational resources** hơn |


## 2. Why Does "bank" Need Contextual Representation?

*(Tại sao từ "bank" cần một cách biểu diễn phụ thuộc vào ngữ cảnh?)*

**Question:** Why does the word "bank" need a contextual representation?

(Tại sao từ "bank" cần một contextual representation?)

Từ **"bank"** có thể có nhiều nghĩa tùy thuộc vào **context**.

- **"I deposited money at the bank."**
  → "bank" mang nghĩa **ngân hàng**.

- **"We sat on the bank of the river."**
  → "bank" mang nghĩa **bờ sông**.

Với **Word2Vec**, mỗi từ chỉ có một **static vector**, vì vậy từ "bank" không có một vector riêng cho từng nghĩa.

Ngược lại, **contextual embeddings** tạo representation dựa trên **surrounding context**. Vì vậy, "bank" trong hai câu trên có thể được biểu diễn khác nhau và phản ánh đúng nghĩa của từ trong từng trường hợp.


## 3. Overall Reflection

*(Nhìn lại những gì đã học được trong Lab)*

**Question:** What did you learn from this lab about word representations and word embeddings?
*(Bạn đã học được gì từ Lab này về word representations và word embeddings?)*


Trong Lab này, em học được cách **word representations** phát triển từ các biểu diễn **sparse** như **TF-IDF** và **co-occurrence matrix** sang các biểu diễn **dense** như **Word2Vec**.

Các thí nghiệm cho thấy chất lượng của **word embeddings** phụ thuộc nhiều vào **corpus**, **context window** và **embedding dimension**. Việc tăng **embedding dimension** không phải lúc nào cũng cho kết quả tốt hơn, đặc biệt khi **training corpus** nhỏ.

Em cũng quan sát được rằng **Word2Vec** có thể phát hiện những quan hệ có ý nghĩa như **doctor – physician**, nhưng đôi khi cũng tạo ra những **surprising similarities** như **doctor – at** hoặc **cat – the**. Điều này xảy ra do corpus nhỏ và các từ có thể xuất hiện trong những context giống nhau.

Cuối cùng, em hiểu được một hạn chế quan trọng của **static embeddings**: một từ chỉ có một vector duy nhất dù có thể có nhiều nghĩa. **Contextual embeddings** giải quyết vấn đề này bằng cách tạo representation khác nhau tùy thuộc vào **context** của từ.
