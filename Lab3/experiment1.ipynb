```python
from WordEmbedding import WordEmbedding

# Small corpus used for the co-occurrence experiment
corpus = [
    "the cat eats fish",
    "the dog eats fish",
    "the cat likes milk",
    "the dog likes meat"
]

model = WordEmbedding()
```

```python
# Build the vocabulary once and reuse it for all window sizes
vocabulary, word_to_index = model.build_vocabulary(corpus)

print("Vocabulary size:", len(vocabulary))
print("Vocabulary:", vocabulary)
```

```python
# Compare how different context window sizes affect the co-occurrence matrix
windows = [1, 2, 5]

for window in windows:
    matrix = model.build_cooccurrence_matrix(
        corpus,
        vocabulary,
        word_to_index,
        window=window
    )

    print(f"\nWindow = {window}")
    print("Matrix shape:", matrix.shape)
    print("Non-zero values:", (matrix != 0).sum())
```

```python
# Compare cosine similarity between selected word pairs
word_pairs = [
    ("cat", "dog"),
    ("cat", "fish"),
    ("dog", "fish")
]

for window in windows:
    matrix = model.build_cooccurrence_matrix(
        corpus,
        vocabulary,
        word_to_index,
        window=window
    )

    print(f"\nWindow = {window}")

    for word1, word2 in word_pairs:
        similarity = model.cosine_similarity(
            matrix[word_to_index[word1]],
            matrix[word_to_index[word2]]
        )

        print(f"{word1} - {word2}: {similarity:.4f}")
```
