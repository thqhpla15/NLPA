```python
from gensim.models import Word2Vec

# Corpus for semantic search

corpus = [
    "the doctor works at the hospital",
    "the physician works at the hospital",
    "the doctor treats the patient",
    "the physician treats the patient",
    "the patient visits the hospital",
    "the doctor studies disease",
    "the physician studies disease",
    "the cat eats fish",
    "the dog eats fish",
    "the cat likes milk",
    "the dog likes meat"
]

sentences = [sentence.split() for sentence in corpus]

model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    epochs=10
)

print("Vocabulary size:", len(model.wv))
print("Vector dimension:", model.wv.vector_size)
```

```python
# Semantic search using Word2Vec embeddings

query = "doctor"

similar_words = model.wv.most_similar(
    query,
    topn=5
)

print(f"Query: {query}")
print("\nRelated words:")

for word, similarity in similar_words:
    print(f"{word}: {similarity:.4f}")
```

```python
# Test semantic search with multiple queries

queries = ["doctor", "hospital", "cat"]

for query in queries:
    print(f"\nQuery: {query}")

    similar_words = model.wv.most_similar(
        query,
        topn=5
    )

    for word, similarity in similar_words:
        print(f"{word}: {similarity:.4f}")
```
