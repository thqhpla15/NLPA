```python
from gensim.models import Word2Vec

corpus = [
    "the doctor works at the hospital",
    "the doctor treats the patient",
    "the patient visits the hospital",
    "the doctor studies disease",
    "the hospital treats many patients",
    "the patient has a disease",
    "the computer is in the hospital",
    "the football player visits the hospital",
    "the doctor uses a computer",
    "the patient uses a computer",
    "the banana is a fruit",
    "the cat eats a banana",
    "the dog eats a banana"
]

sentences = [sentence.split() for sentence in corpus]
```

```python
from gensim.models import Word2Vec

# Train the Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,   # Dimension of each word vector
    window=5,          # Maximum distance between the target word and context words
    min_count=2,       # Ignore words that appear fewer than 2 times
    epochs=10          # Number of training iterations
)

# Check the vocabulary size and vector dimension
print("Vocabulary size:", len(model.wv))
print("Vector dimension:", model.wv.vector_size)
```

```python
words = [
    "doctor",
    "hospital",
    "patient",
    "disease",
    "computer",
    "football",
    "banana"
]

# Check which words are available in the vocabulary
for word in words:
    print(word, "->", word in model.wv)
```

```python
doctor_vector = model.wv["doctor"]

print("Doctor vector:")
print(doctor_vector)
print("Vector shape:", doctor_vector.shape)
```

```python
# Find the words most similar to "doctor"
similar_words = model.wv.most_similar("doctor", topn=5)

print("Words most similar to doctor:")
for word, similarity in similar_words:
    print(f"{word}: {similarity:.4f}")
```

```python
print("Vocabulary:")
print(model.wv.index_to_key)
```

```python
words = [
    "doctor",
    "hospital",
    "patient",
    "disease",
    "computer",
    "football",
    "banana"
]

for word in words:
    print(f"\nMost similar to '{word}':")

    if word not in model.wv:
        print("Not in vocabulary")
        continue

    for similar_word, similarity in model.wv.most_similar(word, topn=3):
        print(f"{similar_word}: {similarity:.4f}")
```
