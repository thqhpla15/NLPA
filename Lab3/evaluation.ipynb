```python
# Word pairs for evaluating embedding quality

word_pairs = [
    ("doctor", "physician"),
    ("car", "automobile"),
    ("king", "queen"),
    ("cat", "dog"),
    ("computer", "banana")
]

print("Word pairs:")
for word1, word2 in word_pairs:
    print(f"{word1} - {word2}")
```

```python
import time
import numpy as np
from gensim.models import Word2Vec
```

```python
# Corpus for Experiment 4

corpus_exp4 = [
    "the doctor works at the hospital",
    "the physician works at the hospital",
    "the doctor treats the patient",
    "the physician treats the patient",
    "the patient visits the hospital",
    "the doctor studies disease",
    "the physician studies disease",
    "the cat eats fish",
    "the dog eats fish",
    "the cat likes milk",
    "the dog likes meat"
]

sentences_exp4 = [sentence.split() for sentence in corpus_exp4]

print("Number of sentences:", len(sentences_exp4))
```

```python
import time
from gensim.models import Word2Vec

dimensions = [50, 100, 300]

models_exp4 = {}
training_times = {}

for dimension in dimensions:
    start_time = time.time()

    model_exp4 = Word2Vec(
        sentences=sentences_exp4,
        vector_size=dimension,
        window=5,
        min_count=1,
        epochs=10
    )

    end_time = time.time()

    models_exp4[dimension] = model_exp4
    training_times[dimension] = end_time - start_time

print("Models recreated:", list(models_exp4.keys()))
```

```python
# Evaluate word-pair similarity across different dimensions

for dimension in dimensions:
    model = models_exp4[dimension]

    print(f"\nDimension = {dimension}")

    for word1, word2 in word_pairs:
        if word1 in model.wv and word2 in model.wv:
            similarity = model.wv.similarity(word1, word2)
            print(f"{word1} - {word2}: {similarity:.4f}")
        else:
            print(f"{word1} - {word2}: Not in vocabulary")
```

```python
# Find the highest similarity pair for each dimension

for dimension in dimensions:
    model = models_exp4[dimension]

    available_pairs = []

    for word1, word2 in word_pairs:
        if word1 in model.wv and word2 in model.wv:
            similarity = model.wv.similarity(word1, word2)
            available_pairs.append((word1, word2, similarity))

    if available_pairs:
        best_pair = max(available_pairs, key=lambda x: x[2])

        print(f"Dimension = {dimension}")
        print(
            f"Highest similarity: "
            f"{best_pair[0]} - {best_pair[1]} = {best_pair[2]:.4f}"
        )
```

```python
# Compare embedding similarity with semantic intuition

expected_relationship = {
    ("doctor", "physician"): "High",
    ("car", "automobile"): "High",
    ("king", "queen"): "High",
    ("cat", "dog"): "Moderate",
    ("computer", "banana"): "Low"
}

for pair, relationship in expected_relationship.items():
    print(f"{pair[0]} - {pair[1]}: Expected = {relationship}")
```
